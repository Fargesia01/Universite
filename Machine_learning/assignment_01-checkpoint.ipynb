# 2. Error and loss

Understanding the concept of loss is fundamental to comprehending the general idea of an "error". ML is founded on the basis of recognizing error and minimizing it.

#### 2.1 **[3pt]** About the *Loss Function*: why $L(\hat y, y) = 0 \;\;\text{if}\;\; \hat y = y, \;\;\forall y \in Y$?

Remember that the *Loss* is a simple difference between prediction and target.

The loss function $L$ measures how far off the prediction is from the truth.

When the prediction $\hat{y}$ equals the target $y$, there’s no error—the model got it exactly right. If you plug $\hat{y} = y$ into the loss function:

For absolute difference: $|\hat{y} - y| = |y - y| = |0| = 0$,
For squared difference: $(\hat{y} - y)^2 = (y - y)^2 = 0^2 = 0$.
In both cases, the loss becomes zero because there’s no difference to measure. This makes sense: if the model predicts perfectly, the “error” should be nothing. This property holds for all possible values of $y$ in the target space $Y$, as long as the loss function is designed to be zero when prediction matches reality—which is a fundamental idea in machine learning. The goal is to minimize loss, and a loss of zero is the ideal scenario.

#### 3.4 **[1pt]** Can Logistic Regression be used to predict continuous numerical values like Linear Regression? Why or why not?

No, logistic regression cannot be used to predict continuous numerical values like linear regression does.

Linear regression is designed to predict continuous outcomes—like house prices, temperatures, or heights—where the output can take any value within a range. It models the relationship between inputs and the output as a straight line (or a hyperplane in higher dimensions), producing a number that can vary smoothly across an infinite range.

Logistic regression, on the other hand, is built for classification tasks, not continuous prediction. It predicts the probability of an input belonging to a specific category—like whether an email is spam (yes/no) or a patient has a disease (positive/negative). Its output is a value between 0 and 1, interpreted as a probability, which is then thresholded to assign a class (e.g., if probability > 0.5, predict "yes"). This is done using the logistic (sigmoid) function, which squashes the linear combination of inputs into that 0-to-1 range.

````python
def dec_tree(features):
  if features[2] == 20:
    if features[0] < 500:
      return "small cat"
    elif features[0] > 1000:
      return "big cat"
    else:
      return "normal cat"

  elif features[2] == 18:
    if features[0] < 500:
      return "small dog"
    elif features[0] > 1500:
      return "big dog"
    else:
      return "normal dog"

predict = lambda data: [dec_tree(f_list) for f_list in data]
predictions = predict(pets)
````
